```python
orders = [
    {"order_id": 101, "restaurant": "Spice Route", "status": " Delivered", "amount": 240},
    {"order_id": 102, "restaurant": "Green Bowl", "status": "pending", "amount": 180},
    {"order_id": 103, "restaurant": "Spice Route", "status": "DELIVERED ", "amount": 310},
    {"order_id": 104, "restaurant": "Tandoor Hub", "status": "cancelled", "amount": 150},
    {"order_id": 105, "restaurant": "Green Bowl", "status": "Late", "amount": 95},
    {"order_id": 106, "restaurant": "Tandoor Hub", "status": "delivered", "amount": 205},
    {"order_id": 107, "restaurant": "Spice Route", "status": "  Pending", "amount": 120},
]


def status_revenue(orders, status="delivered"):
    # Clean the requested status once (outer spaces removed, lowercase)
    target = status.strip().lower()

    # Running total starts at 0
    total = 0

    # Loop over the orders, clean each status the same way,
    # and add the amount when the two statuses match
    for order in orders:
        if order["status"].strip().lower() == target:
            total += order["amount"]

    # Return the total (no printing inside the function)
    return total


print(status_revenue(orders))
print(status_revenue(orders, status="pending"))
print(status_revenue(orders, status=" Cancelled "))


# Bonus: revenue for every cleaned status, reusing status_revenue
def revenue_by_status(orders):
    cleaned_statuses = {order["status"].strip().lower() for order in orders}
    return {s: status_revenue(orders, status=s) for s in cleaned_statuses}


print(revenue_by_status(orders))
```

```python
import csv
import json

# Setup: creates the sample files. Already filled in for you.
sample_orders = [
    {"order_id": 301, "customer": "C-11", "restaurant": "Spice Hub", "items": "Paneer Wrap", "amount": 180},
    {"order_id": 302, "customer": "C-12", "restaurant": "Green Bowl", "items": "Salad", "amount": 220},
    {"order_id": 303, "customer": "C-13", "restaurant": "Spice Hub", "items": "Dal Rice", "amount": 150},
]
with open("week_orders.json", "w", encoding="utf-8") as f:
    json.dump(sample_orders, f, indent=2)
with open("broken_orders.json", "w", encoding="utf-8") as f:
    f.write('[{"order_id": 304 "customer": "C-14"}]')


def export_orders(json_name, csv_name):
    try:
        # Try to load JSON data
        with open(json_name, "r", encoding="utf-8") as f:
            orders = json.load(f)
    except FileNotFoundError:
        print(f"Could not find {json_name}. Nothing exported.")
        return 0
    except json.JSONDecodeError:
        print(f"{json_name} is not in a valid JSON format. Nothing exported.")
        return 0
    except Exception:
        print("Something went wrong.")
        return 0

    # Write to CSV
    with open(csv_name, "w", encoding="utf-8", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=orders[0].keys())
        writer.writeheader()
        writer.writerows(orders)
    return len(orders)


# Checks: keep these lines unchanged.
print(export_orders("week_orders.json", "week_orders.csv"))
print(export_orders("broken_orders.json", "broken_orders.csv"))
print(export_orders("april_orders.json", "april_orders.csv"))
with open("week_orders.csv", "r", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
print(rows[0]["order_id"], rows[-1]["amount"])
```

```python
Python Data Structures..
1. List
- Stores multiple values in a single variable.
- We can add, remove, and modify elements.
- Allows duplicate values.
2. Tuple
- Stores multiple values in a single variable.
- We cannot modify elements after creation.
- Allows duplicate values.
3. Set
- Stores unique values (no duplicates).
- Does not maintain a fixed order.
- We can add and remove elements.
4. Dictionary
- Stores data in key-value pairs.
- Each key must be unique.
- We can add, remove, and modify values.
Easy way to remember:
- List – Changeable data.
- Tuple – Unchangeable data.
- Set – Unique data.
- Dictionary – Key-value data.
```

```python
VISITS = [
    ("C01", "Spice Route", 520),
    ("C02", "Dosa Corner", 180),
    ("C03", "Biryani House", 640),
    ("C01", "Dosa Corner", 250),
    ("C04", "Biryani House", 310),
    ("C03", "Spice Route", 150),
    ("C05", "Spice Route", 90),
    ("C04", "Dosa Corner", 420),
    ("C06", "Biryani House", 275),
    ("C02", "Spice Route", 330),
]

def build_segments(visits, min_amount):
    spice = set()
    dosa = set()

    # Loop over visits and add qualifying customers
    for customer, restaurant, amount in visits:
        if restaurant == "Spice Route" and amount >= min_amount:
            spice.add(customer)
        if restaurant == "Dosa Corner" and amount >= min_amount:
            dosa.add(customer)

    # Build segments
    both = sorted(spice & dosa)
    either = sorted(spice | dosa)
    spice_only = sorted(spice - dosa)

    return {
        "both": both,
        "either": either,
        "spice_only": spice_only
    }

if __name__ == "__main__":
    segments = build_segments(VISITS, 200)
    for name, customers in segments.items():
        print(f"{name}: {customers}")
```

```python
import sqlite3

conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

cursor.execute("CREATE TABLE clinics (clinic_id INTEGER PRIMARY KEY, name TEXT, speciality TEXT)")
cursor.execute("CREATE TABLE appointments (appointment_id INTEGER PRIMARY KEY, patient_code TEXT, clinic TEXT)")

clinic_rows = [
    (1, "City Care", "Dental"),
    (2, "Green Leaf", "Skin"),
    (3, "Bone Well", "Ortho"),
    (4, "Heart Point", "Cardio"),
]

appointment_rows = [
    (201, "P01", "City Care"),
    (202, "P02", "Bone Well"),
    (203, "P03", "Sun Clinic"),
    (204, "P04", "Green Leaf"),
    (205, "P05", "City Care"),
    (206, "P06", "Lake View"),
    (207, "P07", "Bone Well"),
]

for row in clinic_rows:
    cursor.execute("INSERT INTO clinics VALUES (?, ?, ?)", row)
for row in appointment_rows:
    cursor.execute("INSERT INTO appointments VALUES (?, ?, ?)", row)

# ✅ Query to find appointments whose clinic is missing from clinics table
query = """
SELECT a.appointment_id, a.patient_code, a.clinic
FROM appointments AS a
LEFT JOIN clinics AS c
ON a.clinic = c.name
WHERE c.name IS NULL
ORDER BY a.appointment_id;
"""

for row in cursor.execute(query):
    print(row)




    SELECT orders.order_id,
       orders.customer,
       orders.restaurant,
       orders.amount,
       restaurants.cuisine
FROM orders
LEFT JOIN restaurants
       ON orders.restaurant = restaurants.name
WHERE orders.status = 'delivered'
  AND (restaurants.cuisine <> 'Italian' OR restaurants.cuisine IS NULL)
ORDER BY orders.amount DESC;

Why this works

LEFT JOIN starts from orders, so an order like #104 (Wok Express, which isn’t in restaurants) stays in the result with cuisine = NULL.
restaurants.cuisine <> 'Italian' alone would silently drop NULL rows, because comparing NULL to anything evaluates to unknown. The explicit OR restaurants.cuisine IS NULL keeps them.
The OR is wrapped in parentheses so the status = 'delivered' filter applies to both the matched and unmatched cases. Without them, AND binds tighter than OR, and cancelled orders with a NULL cuisine (like #108) would leak into the output.
The status and amount columns are qualified with orders., and the filter on status goes in WHERE since it only involves the left table.

Result on the sample data

order_id | customer | restaurant    | amount | cuisine
107      | cust_07  | Biryani Point | 510    | Biryani
101      | cust_01  | Tandoor House | 420    | North Indian
104      | cust_04  | Wok Express   | 290    | NULL
103      | cust_03  | Dosa Corner   | 150    | South Indian

Bonus query

sql
SELECT restaurants.cuisine,
       COUNT(*)           AS order_count,
       SUM(orders.amount) AS total_amount
FROM orders
LEFT JOIN restaurants
       ON orders.restaurant = restaurants.name
WHERE orders.status = 'delivered'
  AND (restaurants.cuisine <> 'Italian' OR restaurants.cuisine IS NULL)
GROUP BY restaurants.cuisine
ORDER BY total_amount DESC;

The unmatched orders form their own group with a NULL cuisine, since GROUP BY treats all NULLs as one group.
```
